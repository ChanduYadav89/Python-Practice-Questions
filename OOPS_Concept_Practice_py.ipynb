# ✅✅ OOPS CONCEPTS **😊**.

```python
# class making and object making

class Car:
  def start(self):
    print("Car start")

  def stop(self):
    print("Car stop")

obj = Car()

obj.start()
obj.stop()
```

```python
# __init__ usecase

class Car:

  def __init__(self, brand , color, model):

    self.brand = brand
    self.color = color
    self.model = model




car1 = Car("Mahindra", "Black", "Scorpio Classic")

car2 = Car("BMW", "White", "Gt")

print("Car Model:-",car1.model)

print("Car Brand:- ",car1.brand)

print("Car Color:-",car2.color)
```

```python
# use __init__ as take an arguments

class User:
  def __init__(self, name, marks):
    self.name = name
    self.marks = marks

  def bonus_added(self, bonus):
    self.marks += bonus
    print("Bonus added in marks:-",bonus)

  def show_details(self):
    print("Name:-",self.name)
    print("Marks:-", self.marks)

user1 = User("Chandu Yadav", 80)

user1.bonus_added(5)

user1.show_details()
```
