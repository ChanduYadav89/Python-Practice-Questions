# ✅✅ OOPS CONCEPTS **😊**.

```python
# class making and object making

class Car:
  def start(self):
    print("Car start")

  def stop(self):
    print("Car stop")

obj = Car()

obj.start()
obj.stop()
```

```python
# __init__ usecase

class Car:

  def __init__(self, brand , color, model):

    self.brand = brand
    self.color = color
    self.model = model




car1 = Car("Mahindra", "Black", "Scorpio Classic")

car2 = Car("BMW", "White", "Gt")

print("Car Model:-",car1.model)

print("Car Brand:- ",car1.brand)

print("Car Color:-",car2.color)
```
