# ✅➡️ Custom **Functions** 😊😊

```python
# Question 1: Create a function to add two numbers.

def add(a,b):
  return a+b

add(20,30)
```

```python
# Question 2: Check whether a number is even or odd.

def even_odd(num):
  if num % 2 == 0:
    return "Even"
  else:
    return "Odd"

even_odd(20)

```

```python
# Question 3: Find the largest of three numbers

def largest(num1, num2, num3):
  return print(f"The largest number is:-",max(num1,num2,num3))

largest(20,45,12)
```

```python
# Question 4: Calculate the factorial of a number.

def factorial(num):
  fact = 1
  for i in range(1,num+1):
    fact *= i
  return fact

factorial(5)
```

```python
# Question 5: Create a function with a default parameter.

def details(name, age=18):
  if age > 18:
    return print(f"{name}, is eligible for driving car bc their age:- {age} is greater than 18 years")
  else:
    return print("Not eligible")

details("Chandu Yadav", 20)
```

```python
# Question 6: Calculate the sum of all numbers in a list.

def sum_all(num):
  return sum(num)

sum_all([20,32,45,87,95,65,12,47])
```

```python
# Question 7: Check whether a string is a palindrome

def palindrome(text):
  if text == text[::-1]:
    return True
  else:
    return False

palindrome("Chandu")
```

```python
# Question 8: Count vowels in a string.
def vowels(char):
  return  len([vowel  for vowel in char if vowel.lower() in "aeiou"])

vowels("ChandanE")
```

Task 2: Online Shopping Bill Calculator
An e-commerce company wants to calculate the total cost of an order.

Your Task
Create a function named calculate_order_total() that:

Accepts price and quantity.
Calculates the total order amount.
Returns the calculated amount.
Store the returned value in a variable and display it outside the function.

Expected Formula
Order Total = Price × Quantity
Test Cases
Price: 500, Quantity: 3
Expected Total: 1500

Price: 1299, Quantity: 2
Expected Total: 2598

```python
def bill_calc(price, quantity):

  total_order = (price * quantity)

  return total_order

bill_calc(500,3)
```

# ➡️ Lambda function with Map.

```python
# Square the number.
numbers = [45,21,20,1,36,25,47,58,35]

square = list(map(lambda num : num**2 , numbers))  # Here, I can directly use the lambda function in a map function.

print(square)
```
