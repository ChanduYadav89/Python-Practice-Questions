# ✅ Basic List **Comprehension**

```python
# Q1. Find Even Squares numbers from 1 to 20

Even = [num**2 for num in range(1,21) if num % 2 == 0 ]

print(Even)
```

```python
# Q2. Even Numbers
# Create a list of all even numbers from 1 to 50.

Even_num = [num for num in range(1,51) if num % 2 ==0]
print(Even_num)
```

```python
# Q3. Odd Numbers
# Create a list of all odd numbers from 1 to 30.

Odd = [num for num in range(1,31) if num % 2 != 0]
print(Odd)
```

```python
# Double the Numbers
numbers = [2, 4, 6, 8, 10]

Double = [num*num for num in numbers ]
print(Double)
```
