# ✅ Basic List **Comprehension**

```python
# Q1. Find Even Squares numbers from 1 to 20

Even = [num**2 for num in range(1,21) if num % 2 == 0 ]

print(Even)
```

```python
# Q2. Even Numbers
# Create a list of all even numbers from 1 to 50.

Even_num = [num for num in range(1,51) if num % 2 ==0]
print(Even_num)
```

```python
# Q3. Odd Numbers
# Create a list of all odd numbers from 1 to 30.

Odd = [num for num in range(1,31) if num % 2 != 0]
print(Odd)
```

```python
# Q4. Double the Numbers
numbers = [2, 4, 6, 8, 10]

Double = [num*num for num in numbers ]
print(Double)
```

```python
# Q5. Length of Words
# Create a list containing the length of each word.

words = ["Python", "Java", "C++", "JavaScript"]

length = [len(word) for word in words]
print(length)
```

```python
# Q6. Convert Numbers to Negative
numbers = [10, 20, 30, 40, 50]

negative = [(-num) for num in numbers]
print(negative)
```

```python
# Q7. Convert Celsius to Fahrenheit
# Convert each temperature into Fahrenheit
celsius = [0, 10, 20, 30, 40]

faren = [(c*9/5)+32 for c in celsius ]
print(faren)

```

```python
# Q8. Convert Words to Uppercase
# Convert every word to uppercase.
words = ["python", "java", "html", "css"]

upper = [word.upper() for word in words]
print(upper)
```

```python
# Q9. Cube of Numbers
# Create a list containing the cube of every number.

numbers = [1, 2, 3, 4, 5]

cube = [pow(num,3) for num in numbers]
print(cube)
```

```python
# Q10. Extract Characters
# Create a list containing each character separately.

name = "Chandu Yadav"

word = [i for i in name]
print(word)
```
