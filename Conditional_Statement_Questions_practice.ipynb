# **Level 1.  Easy Questions**

```python
# Qu 1. Write a Python program that takes an integer as input and checks whether it is positive, negative, or zero.

num = int(input("Enter a number:- "))

if num > 0:
  print(f"{num} is Positive number")
elif num < 0:
  print(f"{num} is Negative number")
else:
  print(f"{num} is Zero number")
```

```python
# Qu 2. Write a program to check whether a given number is even or odd.

num = int(input("Enter a number:- "))

if num % 2 == 0:
  print(f"{num} is Even Number")
else:
  print(f"{num} is Odd Number")
```

```python
# Qu 3. Take two numbers as input and print the greater number. If both numbers are equal, print "Both are equal".

num1 = int(input("Enter the first number:- "))
num2 = int(input("Enter the second number:- "))

if num1 > num2:
  print(f"{num1} is greater number")
elif num2 > num1:
  print(f"{num2} is greater number")
else:
  print(f"{num1} = {num2} Both are equal number")
```

```python
# Qu 4. Take an integer as input and check whether it is divisible by both 5 and 11.

num = int(input("Enter the number:- "))

if num % 5 == 0 and num % 11 == 0:
  print(num,"is divisible by both 5 and 11")
else:
  print(num , "is not divisible by 5 and 11")
```

```python
# Qu 5. Grade Calculator

# Write a program that assigns grades according to these rules:

# Marks	    Grade
# 90–100	   A
# 80–89	     B
# 70–79	     C
# 60–69     	D
# Below 60	  F

#Also, handle marks below 0 or above 100 as invalid.

marks = int(input("Enter the marks:- "))

if marks <= 0 or marks > 100:
  print(marks, "is invalid Marks, plzz provide correct marks")
elif marks >= 90:
  print(" Grade A")
elif marks >= 80:
  print("Grade B")
elif marks >= 70:
  print("Grade C")
elif marks >= 60:
  print("Grade D")
else:
  print("Grade F")
```
