# ✅✅ SERIES

```python
import pandas as pd

marks = [54,21,35,47,95,84,35,1,0]

student = pd.Series(marks)

print(student)
```

```python
import pandas as pd

marks = [45,21,9,47,56,87]
student = ["Sahil","Shivam","guru","Ram","Shyam","Chandu"]

data = pd.Series(marks , index = student)

print(data)
```

```python
import pandas as pd

student_marks = {
    "Chandu": 85,
    "Sahil": 90,
    "Shivam": 78,
    "Priya": 95
}

data = pd.Series(student_marks)
print(data)

```

```python
print("Data Values:-")
print(data.values)

print("\nData Index:-")
print(data.index)

print("\nSum of values:- ")
print(data.sum())

print("\nAverage of data:- ")
print(data.mean())

print("\nMax of data:- ")
print(data.max())

print("\nMin of Data:- ")
print(data.min())
```
