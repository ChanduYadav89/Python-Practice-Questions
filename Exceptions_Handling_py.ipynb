```python
# Q1. Handle ZeroDivisionError

# Write a Python program that:

# Takes two numbers as input.
# Divides the first number by the second number.
# If the user enters 0 as the second number, handle the ZeroDivisionError.
# Print "Cannot divide by zero" when the exception occurs.

def divide(a,b):
  try:
    return a/b
  except ZeroDivisionError:
    return "Cannot divide by zero"

divide(40,0)
```

```python

# Q2. Handle ValueError (Beginner)

# Write a Python program that:

# Takes input from the user.
# Converts that input into an integer using int().
# If the user enters something like "hello" instead of a number, handle ValueError.
# Print "Please enter a valid integer".


def get_number():
    try:
        num = int(input("Enter a number: "))
        print("Your number is:", num)

    except ValueError:
        print("Please enter a valid integer")


get_number()
```
