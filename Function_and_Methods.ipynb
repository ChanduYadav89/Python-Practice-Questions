```python
# Q1. Sum of Two Numbers

def sum_digit(a,b):
  print(a+b)

sum_digit(45,89)
```

```python
# Qu 2. Even or Odd

def even_odd(a):

  if a % 2 == 0:
    print("Even")
  else:
    print("Odd")

even_odd(89)
```

```python
# Qu 3. Find the Largest Number

def largest(num1, num2, num3):

  print("Largest number is:-",max(num1, num2, num3))

largest(78,89,64)

```

```python
# Q4. Calculate Square

def square(num):
   print(num*num)

square(20)
```

```python
# Q5. Count Characters

def char(chars):

  print(len(chars))

char("Chandu Yadav")
```

```python
# Q6. Positive, Negative or Zero

def number(num):
  if num > 0:
    print(num,"Positive number")
  elif num < 0:
    print(num,"Negative number")
  else:
    print(num,"zero number")

number(88)
```

```python
# Q7. Calculate Factorial

def factorial(num):

  fact = 1

  for num in range(1, num+1):
    fact *= num

  print(fact)

factorial(8)

```

```python
# Q1. Prime Number

def prime(num):
  if num < 2:
    return False

  for i in range(2,num):
    if num % i == 0:
      return False

  return True


num = int(input("Enter a number: "))

if prime(num):
    print("Prime Number")
else:
    print("Not a Prime Number")
```
