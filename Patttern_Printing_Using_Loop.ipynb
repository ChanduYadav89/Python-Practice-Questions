# ✅✅ **Pattern** Printing Practice Questions 😊

```python
# Q1. Print the right-angle triangle with the number of rows as 5.

n = 6

for row in range(n):
  for col in range(row):
    print("*",end="")

  print()
```

```python
# Q2. Print a rhombus where the number of rows is 5.
n = 6

for row in range(n):
  for col in range(n-row):
    print(" ",end="")
  for col in range(n):
    print("*",end="")

  print()
```

```python
# Q3. The inverted right-angle triangle with the number of rows is 5.
n = 5

for row in range(n):
  for col in range(n-row):
    print("*",end="")

  print()
```

```python
# Q4. Print the triangle with the number of rows equal to 5.

n = 5

for row in range(n):
  for col in range(n-row):
    print(" ",end="")

  for col in range(2*row-1):
    print("*",end="")

  print()
```
