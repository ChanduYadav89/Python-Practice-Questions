# ✅✅ **Pattern** Printing Practice Questions 😊

```python
# Q1. Print the right-angle triangle with the number of rows as 5.

n = 6

for row in range(n):
  for col in range(row):
    print("*",end="")

  print()
```

```python
# Q2. Print a rhombus where the number of rows is 5.
n = 6

for row in range(n):
  for col in range(n-row):
    print(" ",end="")
  for col in range(n):
    print("*",end="")

  print()
```

```python
# Q3. The inverted right-angle triangle with the number of rows is 5.
n = 5

for row in range(n):
  for col in range(n-row):
    print("*",end="")

  print()
```

```python
# Q4. Print the triangle with the number of rows equal to 5.

n = 5

for row in range(n):
  for col in range(n-row):
    print(" ",end="")

  for col in range(2*row-1):
    print("*",end="")

  print()
```

```python
# Q5. Print a reverse triangle with the number of rows equal to 4.
n = 4

for row in range(n):
  for col in range(row):
    print(" ",end="")

  for col in range(2*n-2*row-1):
    print("*",end="")

  print()
```

```python
# Qu 6. Print the diamond pattern.
n = 5

# Upper half
for i in range(1, n + 1):
    print(" " * (n - i), end="")
    print("*" * (2 * i - 1))

# Lower half
for i in range(n - 1, 0, -1):
    print(" " * (n - i), end="")
    print("*" * (2 * i - 1))
```
